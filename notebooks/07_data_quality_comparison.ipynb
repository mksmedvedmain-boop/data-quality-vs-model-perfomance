# Comparison of Data Quality Factors

## Introduction

Previous experiments investigated how different aspects of data quality influence the performance of a Logistic Regression model.

Three independent experiments were completed:

- Missing Data
- Label Noise
- Feature Selection

The objective of this notebook is to compare these experiments and identify which data quality factor has the greatest impact on model performance.

## Import Libraries

The following libraries are required to create comparison tables and visualize the results of previous experiments.

```python
import pandas as pd

import matplotlib.pyplot as plt
```

## Create Comparison Table

The best result from each experiment is selected for comparison.

Only the highest-performing version of each experiment is included to evaluate the overall influence of different data quality factors.

```python
comparison = pd.DataFrame({"Experiment": ["Missing Data", "Label Noise","Feature Selection"],"Accuracy": [ 0.804,0.810,0.810 ],"Precision": [0.783,0.786,0.786],"Recall": [0.730, 0.743,0.743], "F1-score": [ 0.755,0.764,0.764]})

comparison
```

## Compare the Results

The comparison table summarizes the best-performing configuration from each experiment.

This allows the influence of different data quality factors to be compared using the same evaluation metrics.

```python
comparison
```

## Accuracy Comparison

The following graph compares the Accuracy obtained in each experiment.

```python
plt.figure(figsize=(8,5))

plt.plot(comparison["Experiment"],comparison["Accuracy"],marker="o")

plt.xlabel("Experiment")
plt.ylabel("Accuracy")
plt.title("Comparison of Accuracy Across Data Quality Experiments")

plt.grid(True)

plt.savefig("comparison_accuracy.png")

plt.show()
```

## F1-score Comparison

The following graph compares the F1-score obtained in each experiment.

The F1-score provides a balanced evaluation by considering both Precision and Recall.

```python
plt.figure(figsize=(8,5))

plt.plot(comparison["Experiment"],comparison["F1-score"],marker="o")

plt.xlabel("Experiment")
plt.ylabel("F1-score")
plt.title("Comparison of F1-score Across Data Quality Experiments")

plt.grid(True)

plt.savefig("comparison_f1.png")

plt.show()
```

## Analysis of Results

The comparison demonstrates that different aspects of data quality influence machine learning performance in different ways.

The Missing Data experiment showed only a small decrease in model performance. Even after introducing missing values, the Logistic Regression model maintained relatively stable Accuracy and F1-score.

The Label Noise experiment produced the largest performance degradation. Incorrect target labels significantly reduced the model's ability to learn meaningful patterns, making label quality the most critical factor among the experiments.

The Feature Selection experiment showed that using different sets of informative features changed Precision, Recall, and F1-score only slightly. The model achieved its best balanced performance when all informative features were included.

Overall, the experiments indicate that machine learning models are considerably more sensitive to incorrect labels than to moderate amounts of missing data or small differences in feature selection.

## Conclusion

This comparative study demonstrates that data quality has a direct impact on machine learning performance.

Among the investigated factors, label noise had the strongest negative influence on the Logistic Regression classifier, while missing data produced only a moderate decrease in performance. Feature selection affected individual evaluation metrics but did not substantially change the overall model performance.

These findings confirm that ensuring correct target labels is one of the most important steps in developing reliable machine learning models. Careful feature selection and handling missing values remain important, but their influence is smaller than the effect of incorrect labels.
