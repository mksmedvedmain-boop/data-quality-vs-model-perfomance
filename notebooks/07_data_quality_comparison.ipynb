# Comparison of Data Quality Factors

## Introduction

Previous experiments investigated how different aspects of data quality influence the performance of a Logistic Regression model.

Three independent experiments were completed:

- Missing Data
- Label Noise
- Feature Selection

The objective of this notebook is to compare these experiments and identify which data quality factor has the greatest impact on model performance.

## Import Libraries

The following libraries are required to create comparison tables and visualize the results of previous experiments.

```python
import pandas as pd

import matplotlib.pyplot as plt
```

## Create Comparison Table

The best result from each experiment is selected for comparison.

Only the highest-performing version of each experiment is included to evaluate the overall influence of different data quality factors.

```python
comparison = pd.DataFrame({ "Data Quality Issue": [ "Missing Data", "Label Noise","Feature Selection"],"10% Accuracy": [ 0.777, 0.469,0.810],"20% Accuracy": [0.765,0.196,0.810],"30% Accuracy": [ 0.765,0.196, 0.810], "10% F1-score": [ 0.730, 0.410, 0.757 ],"20% F1-score": [ 0.724, 0.182, 0.761],"30% F1-score": [ 0.724,0.182, 0.761]})
```

## Compare the Results

The comparison table summarizes the best-performing configuration from each experiment.

This allows the influence of different data quality factors to be compared using the same evaluation metrics.

```python
comparison
```

## Accuracy Comparison

The following graph compares the effect of Missing Data and Label Noise on Accuracy at 10%, 20%, and 30% degradation levels.

This comparison helps identify which type of data quality problem has a stronger effect on Logistic Regression performance.

```python


missing_data_accuracy = [0.777, 0.765, 0.765]
label_noise_accuracy = [0.469, 0.196, 0.196]

missing_levels = ["10%", "20%", "30%"]

plt.figure(figsize=(8,5))

plt.plot(missing_levels, missing_data_accuracy, marker="o",label="Missing Data")

plt.plot(missing_levels,label_noise_accuracy,marker="o",label="Label Noise")

plt.xlabel("Data Degradation Level")
plt.ylabel("Accuracy")
plt.title("Impact of Missing Data and Label Noise on Accuracy")

plt.grid(True)
plt.legend()

plt.savefig("missing_vs_label_noise_accuracy.png")

plt.show()
```

## F1-score Comparison

The following graph compares the F1-score of Logistic Regression under different levels of Missing Data and Label Noise.

F1-score is used to provide a balanced comparison of Precision and Recall.

```python
missing_data_f1 = [0.730, 0.724, 0.724]
label_noise_f1 = [0.410, 0.182, 0.182]

plt.figure(figsize=(8,5))

plt.plot(missing_levels, missing_data_f1, marker="o",label="Missing Data")

plt.plot(missing_levels, label_noise_f1,marker="o",label="Label Noise")

plt.xlabel("Data Degradation Level")
plt.ylabel("F1-score")
plt.title("Impact of Missing Data and Label Noise on F1-score")

plt.grid(True)
plt.legend()

plt.savefig("missing_vs_label_noise_f1.png")

plt.show()
```

## Feature Selection Accuracy

The following graph compares Accuracy when different feature sets are used.

The experiment evaluates whether removing or adding features substantially changes the model's overall predictive performance.

```python
feature_sets = ["Basic Features","Extended Features","All Features"]

feature_accuracy = [0.810, 0.810, 0.810]

plt.figure(figsize=(8,5))

plt.plot( feature_sets,feature_accuracy, marker="o")

plt.xlabel("Feature Set")
plt.ylabel("Accuracy")
plt.title("Accuracy Across Different Feature Sets")

plt.grid(True)

plt.savefig("feature_selection_accuracy.png")

plt.show()
```

## Feature Selection F1-score

The following graph compares the F1-score across the different feature configurations.

This helps determine whether feature selection affects the balance between Precision and Recall.

```python
feature_f1 = [0.761, 0.757, 0.764]

plt.figure(figsize=(8,5))

plt.plot(feature_sets,feature_f1,marker="o")

plt.xlabel("Feature Set")
plt.ylabel("F1-score")
plt.title("F1-score Across Different Feature Sets")

plt.grid(True)

plt.savefig("feature_selection_f1.png")

plt.show()
```

## Analysis of Results

The results demonstrate that different data quality factors have very different effects on Logistic Regression performance.

Label Noise had the strongest negative impact on the model. Accuracy decreased from 0.810 with no label noise to 0.469 at 10% noise and to 0.196 at both 20% and 30% noise. The F1-score followed the same pattern, decreasing from 0.764 to 0.410 at 10% noise and to 0.182 at 20% and 30% noise.

Missing Data had a smaller but still noticeable effect. Accuracy decreased from 0.810 with the original data to 0.777 at 10% missing data and 0.765 at 20% and 30%. The F1-score decreased from 0.764 to 0.730 at 10% and 0.724 at 20% and 30%.

Feature Selection had the smallest effect on overall performance. Accuracy remained at 0.810 for Basic, Extended, and All Features. The F1-score changed only slightly, ranging from 0.757 to 0.764.

Overall, the results show that Logistic Regression was much more sensitive to incorrect target labels than to missing values or changes in the selected features. Label Noise caused a substantial degradation in performance, while Missing Data caused a moderate decrease and Feature Selection produced only minor changes.

## Conclusion

In this experiment, three data quality factors were compared to determine their influence on Logistic Regression performance: Missing Data, Label Noise, and Feature Selection.

Label Noise had the strongest negative effect on the model. Even 10% label noise caused a substantial decrease in Accuracy and F1-score, while higher levels of noise reduced the model's performance even further.

Missing Data also reduced model performance, but the effect was considerably smaller. Accuracy decreased moderately and then remained stable between 20% and 30% missing data.

Feature Selection had the smallest influence. Accuracy remained unchanged across the tested feature configurations, while Precision, Recall, and F1-score changed only slightly.

These findings support the hypothesis that data quality has a significant influence on machine learning performance. In particular, the correctness of target labels appears to be more important than moderate missing data or differences in feature selection for the Logistic Regression model.
