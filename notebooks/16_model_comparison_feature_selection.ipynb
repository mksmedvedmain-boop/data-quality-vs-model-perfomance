# Comparison of Machine Learning Models Under Feature Selection

## Introduction

Previous experiments investigated how feature selection influences the performance of individual machine learning models.

Three independent experiments have now been completed:

- Logistic Regression
- Decision Tree
- Random Forest

The objective of this notebook is to compare the performance of these three algorithms under different feature selection configurations and determine which model is most robust when different feature sets are used.

## Import Libraries

The following libraries are required to create comparison tables and visualize the results of the previous experiments.

```python
import pandas as pd

import matplotlib.pyplot as plt
```

## Create Accuracy Comparison Table

The Accuracy values obtained from the previous feature selection experiments are combined into a single comparison table.

The table presents the Accuracy of Logistic Regression, Decision Tree, and Random Forest for each feature configuration.

This comparison allows the influence of feature selection on different machine learning algorithms to be evaluated.

```python
accuracy_comparison = pd.DataFrame({"Feature Set": ["Basic Features","Extended Features","All Features"],"Logistic Regression Accuracy": [0.810,0.810,0.810],"Decision Tree Accuracy": [ 0.631, 0.749, 0.782 ],"Random Forest Accuracy": [ 0.721,0.810,0.821]})

accuracy_comparison
```

## Accuracy Comparison

The following graph compares the Accuracy of Logistic Regression, Decision Tree, and Random Forest across the three feature selection configurations.

This visualization demonstrates how the overall classification performance changes when different feature sets are used.

```python
plt.figure(figsize=(8,5))

plt.plot(accuracy_comparison["Feature Set"],accuracy_comparison["Logistic Regression Accuracy"],marker="o",label="Logistic Regression")

plt.plot(accuracy_comparison["Feature Set"],accuracy_comparison["Decision Tree Accuracy"],marker="o", label="Decision Tree")

plt.plot(accuracy_comparison["Feature Set"], accuracy_comparison["Random Forest Accuracy"],marker="o",label="Random Forest")

plt.xlabel("Feature Set")
plt.ylabel("Accuracy")
plt.title("Model Comparison Under Feature Selection")

plt.legend()

plt.grid(True)

plt.savefig("model_comparison_feature_selection_accuracy.png")

plt.show()
```

## Create F1-score Comparison Table

The following table presents the F1-score obtained by Logistic Regression, Decision Tree, and Random Forest for each feature configuration.

The F1-score is used to evaluate the balance between Precision and Recall and provides an additional comparison of model performance.

```python
f1_comparison = pd.DataFrame({"Feature Set":["Basic Features","Extended Features","All Features"],"Logistic Regression F1-score":[0.761,0.757,0.764],"Decision Tree F1-score":[0.522, 0.698,0.742],"Random Forest F1-score":[ 0.643,0.764, 0.778]})

f1_comparison
```

## F1-score Comparison

The following graph compares the F1-score of the three machine learning models across different feature selection configurations.

The F1-score provides a balanced evaluation of Precision and Recall.

```python
plt.figure(figsize=(8,5))

plt.plot(f1_comparison["Feature Set"],f1_comparison["Logistic Regression F1-score"],marker="o",label="Logistic Regression")

plt.plot(f1_comparison["Feature Set"],f1_comparison["Decision Tree F1-score"],marker="o",label="Decision Tree")

plt.plot(f1_comparison["Feature Set"],f1_comparison["Random Forest F1-score"],marker="o",label="Random Forest")

plt.xlabel("Feature Set")
plt.ylabel("F1-score")
plt.title("Model Comparison Under Feature Selection")

plt.legend()

plt.grid(True)

plt.savefig("model_comparison_feature_selection_f1.png")

plt.show()
```

## Analysis of Results

The comparison shows that feature selection influenced the three machine learning models differently.

Logistic Regression demonstrated stable performance across all feature configurations, maintaining an Accuracy of 0.810 while showing only minor changes in F1-score. This suggests that the model was relatively insensitive to the selected feature set.

Decision Tree showed the greatest dependence on feature selection. The Basic Features configuration produced the lowest performance, while the Extended Features and All Features configurations substantially improved both Accuracy and F1-score.

Random Forest achieved the strongest overall performance. Similar to Decision Tree, its performance improved as more informative features were included, reaching the highest Accuracy (0.821) and the highest F1-score (0.778) when all processed features were used.

Overall, the results indicate that tree-based models benefit more from richer feature sets, whereas Logistic Regression remains comparatively stable under different feature selection configurations.

## Conclusion

This experiment compared the influence of feature selection on three machine learning algorithms using the Titanic dataset.

Logistic Regression produced stable performance across all feature configurations, while Decision Tree and Random Forest showed noticeable improvements as more informative features were included.

Among the evaluated models, Random Forest achieved the highest overall Accuracy and F1-score, followed by Logistic Regression and Decision Tree.

These findings demonstrate that feature selection affects machine learning algorithms differently and should be considered when selecting the most appropriate model for a classification problem.
