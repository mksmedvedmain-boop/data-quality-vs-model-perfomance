# Comparison of Machine Learning Models Under Label Noise

## Introduction

Previous experiments investigated the effect of label noise on three different machine learning models: Logistic Regression, Decision Tree, and Random Forest.

Each model was evaluated under different levels of label noise using the Titanic dataset.

The objective of this notebook is to compare the performance of all three models under the same levels of label noise and determine which model is the most robust to incorrect target labels.

## Import Libraries

The following libraries are used to combine the results from the previous experiments and visualize the performance of the three machine learning models.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## Results

The results from the Logistic Regression, Decision Tree, and Random Forest label noise experiments are combined into a single comparison table.

The models were evaluated at 0%, 10%, 20%, and 30% label noise using Accuracy, Precision, Recall, and F1-score.

```python
comparison = pd.DataFrame({
        "Model": [
        "Logistic Regression", "Logistic Regression", "Logistic Regression", "Logistic Regression",
        "Decision Tree", "Decision Tree", "Decision Tree", "Decision Tree",
        "Random Forest", "Random Forest", "Random Forest", "Random Forest"],
        "Label Noise": [
        "0%", "10%", "20%", "30%",
        "0%", "10%", "20%", "30%",
        "0%", "10%", "20%", "30%"],

        "Accuracy": [
        0.810, 0.469, 0.196, 0.196,
        0.782, 0.531, 0.223, 0.352,
        0.821, 0.480, 0.184, 0.263],

        "Precision": [
        0.786, 0.471, 0.229, 0.225,
        0.727, 0.545, 0.286, 0.434,
        0.800, 0.486, 0.214, 0.315],

        "Recall": [
        0.743, 0.363, 0.151, 0.152,
        0.757, 0.462, 0.208, 0.343,
        0.757, 0.374, 0.142, 0.219],

        "F1-score": [
        0.764, 0.410, 0.182, 0.182,
        0.742, 0.500, 0.240, 0.383,
        0.778, 0.422, 0.170, 0.258]})

comparison
```

## Accuracy Comparison

The following graph compares the Accuracy of Logistic Regression, Decision Tree, and Random Forest under different levels of label noise.

The graph helps identify how the overall classification performance of each model changes as the quality of the target labels decreases.

```python
plt.figure(figsize=(8, 5))

for model in comparison["Model"].unique():
    data = comparison[comparison["Model"] == model]

    plt.plot(data["Label Noise"],data["Accuracy"],marker="o", label=model)

plt.xlabel("Label Noise")
plt.ylabel("Accuracy")
plt.title("Comparison of Model Accuracy Under Label Noise")

plt.grid(True)
plt.legend()

plt.savefig("model_accuracy_label_noise.png")

plt.show()
```

## F1-score Comparison

The following graph compares the F1-score of the three machine learning models under different levels of label noise.

The F1-score is used because it provides a balanced measure of Precision and Recall.

```python
plt.figure(figsize=(8, 5))

for model in comparison["Model"].unique():
    data = comparison[comparison["Model"] == model]

    plt.plot( data["Label Noise"],data["F1-score"],marker="o", label=model)

plt.xlabel("Label Noise")
plt.ylabel("F1-score")
plt.title("Comparison of Model F1-score Under Label Noise")

plt.grid(True)
plt.legend()

plt.savefig("model_f1_label_noise.png")

plt.show()
```

## Analysis of Results

The comparison shows that all three machine learning models are strongly affected by increasing levels of label noise.

With 0% label noise, Random Forest achieved the highest performance, with an Accuracy of 0.821 and an F1-score of 0.778. Logistic Regression followed with an Accuracy of 0.810 and an F1-score of 0.764, while Decision Tree achieved an Accuracy of 0.782 and an F1-score of 0.742.

When 10% label noise was introduced, Decision Tree achieved the highest Accuracy of 0.531 and the highest F1-score of 0.500. Random Forest achieved an Accuracy of 0.480 and an F1-score of 0.422, while Logistic Regression achieved an Accuracy of 0.469 and an F1-score of 0.410.

At 20% label noise, the performance of all three models decreased substantially. Decision Tree maintained the highest Accuracy of 0.223 and the highest F1-score of 0.240. Logistic Regression achieved an Accuracy of 0.196 and an F1-score of 0.182, while Random Forest showed the lowest performance with an Accuracy of 0.184 and an F1-score of 0.170.

At 30% label noise, Decision Tree again achieved the highest performance, with an Accuracy of 0.352 and an F1-score of 0.383. Random Forest achieved an Accuracy of 0.263 and an F1-score of 0.258, while Logistic Regression remained at an Accuracy of 0.196 and an F1-score of 0.182.

Overall, Decision Tree demonstrated the greatest robustness to label noise among the three models in these experiments. Random Forest achieved the best performance on clean data but experienced a substantial decrease when label noise was introduced.

These results demonstrate that a model that performs best on clean data is not necessarily the most robust when the quality of the target labels decreases.

## Conclusion

In this experiment, Logistic Regression, Decision Tree, and Random Forest were compared under different levels of label noise using the Titanic dataset.

Random Forest achieved the highest performance on the clean dataset, with an Accuracy of 0.821 and an F1-score of 0.778. However, its performance decreased significantly as label noise increased.

Decision Tree demonstrated the strongest overall robustness to label noise. It achieved the highest Accuracy and F1-score at 10%, 20%, and 30% label noise among the three models.

Logistic Regression also experienced a substantial decrease in performance, with Accuracy falling from 0.810 with clean data to 0.196 at both 20% and 30% label noise.

Overall, the experiment demonstrates that label noise can have a strong negative effect on machine learning performance. The results also show that different algorithms respond differently to poor-quality target labels, making robustness an important factor when selecting a machine learning model.
